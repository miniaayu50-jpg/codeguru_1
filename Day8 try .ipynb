```python
def avg(*x):
    sum=0
    num=0
    for i in x:
        sum=sum+i
        num=num+1
    print('Average',sum/num)
```

```python
def avg(*x):
    sum=0
    num=0
    for i in x:
        if type(i)==str:
            continue
        sum=sum+i
        num=num+1
    print('Average',sum/num)
```

```python
def calc(op,no1,no2):
    if(op=="+"):
        print(no1+no2)
    elif(op=="-"):
        print(no1-no2)
    elif(op=="*"):
        print(no1*no2)
    elif(op=="/"):
        print(no1/no2)
calc("/",2,4)
```
