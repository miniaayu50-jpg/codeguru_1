```python
no1 = int(input("Enter first no"))              
no2 = int(input("Enter second no"))
for no in range (no1,no2+1):

    # Not Prime

    for i in range(2,no):
        if no%i==0:
            break
    else:
        continue   
    print('not prime no:',no)
    
    # prime  factors

    for i in range(2,no+1):
        if (no%i==0):

            # Check i is prime
            for j in range(2,i):
                if i%j==0:
                    break
            else:
                print('prime Factor: =',i)
```
