```python
Qut=int(input("Enter the Quntity of items: "))
Rate=int(input("Enter the Rate of items: "))
GST=int(input("Enter the GST percentage: "))

def gstamt(Qut, Rate, GST):
    total_amount = Qut * Rate
    gst_amount = (total_amount * GST) / 100
    final_amount = total_amount + gst_amount
    return final_amount

print("Final Amount (including GST):", gstamt(Qut, Rate, GST))
```
